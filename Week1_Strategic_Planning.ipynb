## 6. Analytical Methodology

1. **Descriptive analytics** – counts, averages, medians, distributions, and grouped KPIs.
2. **Diagnostic analysis** – investigation of relationships between shipping, product, geographic, and operational variables.
3. **Visualization** – charts for trends, distributions, comparisons, and relationships.
4. **Regression/classification** – predictive modeling for a selected logistics outcome.
5. **Correlation analysis** – identification of numerical associations while recognizing that correlation does not prove causation.
6. **Optimization-oriented analysis** – translating validated findings into possible resource allocation, routing, capacity, or process improvements.

## 8. Data Quality and Leakage Considerations

Data quality will be checked before analysis. The project will preserve the raw dataset and create a separate processed version. Missing values, duplicates, invalid values, inconsistent categories, and outliers will be investigated.

For predictive modeling, variables that reveal an outcome only after shipment completion must not be used as prediction inputs when the intended prediction is supposed to happen before that outcome. This prevents data leakage and makes the evaluation more realistic.
